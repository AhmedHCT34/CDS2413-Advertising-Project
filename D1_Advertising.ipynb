```python
import pandas as pd

df = pd.read_csv("advertising.csv")

print(df.head())
```

```python
def descriptive_stats(sample, field):
    data = sample[field]

    print("Mean:", data.mean())
    print("Median:", data.median())
    print("Mode:", data.mode()[0])
    print("Range:", data.max() - data.min())
    print("Variance:", data.var())
    print("Standard Deviation:", data.std())
    print("Minimum:", data.min())
    print("Maximum:", data.max())
    print("Count:", data.count())

descriptive_stats(df, "Sales")
```

```python
random_sample = df.sample(n=150, random_state=42)

print("Random Sample Size:", len(random_sample))

descriptive_stats(random_sample, "Sales")
```

```python
import numpy as np


positions = np.arange(0, len(df), 4)

systematic_sample = df.iloc[positions]

print("Systematic Sample Size:", len(systematic_sample))

descriptive_stats(systematic_sample, "Sales")
```

```python
print("Descriptive Statistical Report for Sales")

print(df["Sales"].describe())

print("Mode:", df["Sales"].mode()[0])
print("Variance:", df["Sales"].var())
print("Range:", df["Sales"].max() - df["Sales"].min())
print("Skewness:", df["Sales"].skew())
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["TV"], df["Sales"])

plt.xlabel("TV Advertising")
plt.ylabel("Sales")
plt.title("TV Advertising vs Sales")

plt.show()
```

```python
plt.boxplot(df["Sales"])

plt.ylabel("Sales")
plt.title("Box Plot of Sales")

plt.show()
```

```python
plt.hist(df["Sales"], bins=10, edgecolor="black")

plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("Histogram of Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

correlation = df.corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
from scipy import stats

t_stat, p_value = stats.ttest_1samp(df["Sales"], 15)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```
